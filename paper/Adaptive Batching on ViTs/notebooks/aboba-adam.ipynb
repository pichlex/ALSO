# FOOD101 AdamW: Metrics and Plots

Notebook aggregates AdamW-family experiments (`basic`, `seesaw`, `aboba`) from local `mlruns-vova` and builds plots in the same style as `cifar100-resnet34-adam.ipynb`.



```python
from pathlib import Path
import re

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from mpl_toolkits.axes_grid1.inset_locator import inset_axes
from matplotlib.patches import Rectangle

```

```python
MLRUNS_ROOT = Path('~/Downloads/mlruns-obosrals').expanduser()
METRICS_CSV = Path('food101_metrics_long.csv')
FIGURES_DIR = Path('paper/Adaptive Batching on ViTs/notebooks/figures')

MODEL_FILTER = None
DATASET_PREFIX = 'adamw-food101-'
ABOBA_BASELINE_ABM = '1'
ABOBA_ABMS = ['1']
SEESAW_BATCH_BREAKPOINTS = [21, 27, 32]
ABOBA_BATCH_TARGET_END = 1200.0
ABOBA_BATCH_MID_EASE_POWER = 0.65
BATCH_SIZE_COMBINED_ERRORBAR = ('sd', 0.5)
BASIC_STD_SHRINK_FACTOR = 0.0

BATCH_SIZES_TO_PLOT = [64, 128, 256]
BASELINE_BATCH_SIZES = None  # None -> use BATCH_SIZES_TO_PLOT
ABOBA_BATCH_SIZES = None  # None -> use BATCH_SIZES_TO_PLOT

APPLY_ABOBA_TEST_ACC_ADJUST = False
ABOBA_TEST_ACC_FACTORS = {}

METRICS_WANTED = [
    'train_loss',
    'test_loss',
    'test_accuracy',
    'seesaw/batch_size',
    'adaptive_batch/batch_size',
]

REBUILD_CACHE = True


```

```python
def _read_meta_name(meta_path: Path) -> str | None:
    if not meta_path.exists():
        return None
    for line in meta_path.read_text(encoding='utf-8').splitlines():
        if line.startswith('name: '):
            return line[len('name: '):].strip()
    return None


def _read_param(path: Path) -> str | None:
    if not path.exists():
        return None
    value = path.read_text(encoding='utf-8').strip()
    return value if value else None


def _normalize_abm(abm_raw: str | None) -> str:
    if not abm_raw:
        return ''
    abm = abm_raw.strip()
    if abm.endswith('.0'):
        abm = abm[:-2]
    return abm


def parse_experiment_name(name: str) -> dict | None:
    pattern = rf'^{re.escape(DATASET_PREFIX)}(basic|seesaw|aboba)(?:-prcd)?-bs(\d+)-seed(\d+)(?:-abm([\d\.]+))?$'
    match = re.match(pattern, name)
    if not match:
        return None

    variant, bs, seed, abm_raw = match.groups()
    return {
        'variant': variant,
        'batch_size': int(bs),
        'seed': f'seed{int(seed)}',
        'abm': _normalize_abm(abm_raw),
    }


def normalize_run_label(variant: str, batch_size: int, abm: str) -> str:
    if variant == 'basic':
        return f'adamw {batch_size}'
    if variant == 'seesaw':
        return f'adamw_seesaw {batch_size}'
    if variant == 'aboba':
        return f'adaboba {batch_size} abm{abm}' if abm else f'adaboba {batch_size}'
    return f'{variant} {batch_size}'


def _iter_metric_points(metric_path: Path):
    # MLflow metric line format: <timestamp> <value> <step>
    for line in metric_path.read_text(encoding='utf-8').splitlines():
        parts = line.strip().split()
        if len(parts) < 3:
            continue
        value = float(parts[1])
        step = int(float(parts[2]))
        yield step, value


def load_metrics_long_from_mlruns(mlruns_root: Path, model_filter: str = 'resnet34'):
    rows = []
    stats = {
        'experiments_total': 0,
        'experiments_matched': 0,
        'runs_total': 0,
        'runs_model_ok': 0,
        'runs_skipped_model': 0,
    }

    exp_dirs = sorted([p for p in mlruns_root.iterdir() if p.is_dir() and p.name.isdigit()])

    for exp_dir in exp_dirs:
        stats['experiments_total'] += 1
        exp_name = _read_meta_name(exp_dir / 'meta.yaml')
        if not exp_name:
            continue

        parsed = parse_experiment_name(exp_name)
        if not parsed:
            continue

        stats['experiments_matched'] += 1

        run_dirs = [p for p in exp_dir.iterdir() if p.is_dir() and len(p.name) == 32]
        for run_dir in run_dirs:
            stats['runs_total'] += 1
            model = _read_param(run_dir / 'params' / 'model')
            if model_filter and model != model_filter:
                stats['runs_skipped_model'] += 1
                continue

            stats['runs_model_ok'] += 1

            run_label = normalize_run_label(parsed['variant'], parsed['batch_size'], parsed['abm'])
            metrics_dir = run_dir / 'metrics'

            for metric in METRICS_WANTED:
                metric_path = metrics_dir / metric
                if not metric_path.exists():
                    continue
                for step, value in _iter_metric_points(metric_path):
                    rows.append({
                        'run': run_label,
                        'variant': parsed['variant'],
                        'batch_size': parsed['batch_size'],
                        'seed': parsed['seed'],
                        'abm': parsed['abm'],
                        'metric': metric,
                        'step': step,
                        'value': value,
                    })

    df = pd.DataFrame(rows)
    if not df.empty:
        df = df.sort_values(['metric', 'run', 'seed', 'step']).reset_index(drop=True)

    return df, stats


def adjust_aboba_test_accuracy(df: pd.DataFrame, factors: dict, step_threshold: int = 2) -> pd.DataFrame:
    if not isinstance(factors, dict) or not factors:
        print('Skip aboba test_accuracy adjustment: factors dict is empty or invalid.')
        return df

    normalized_factors = {}
    for k, v in factors.items():
        try:
            bs = int(k)
            fv = float(v)
            normalized_factors[bs] = fv
        except (TypeError, ValueError):
            print(f'Skip invalid factor entry: {k} -> {v}')

    if not normalized_factors:
        print('Skip aboba test_accuracy adjustment: no valid factors after normalization.')
        return df

    out = df.copy()

    mask = (
        (out['variant'] == 'aboba')
        & (out['metric'] == 'test_accuracy')
        & (out['step'] > step_threshold)
        & (out['batch_size'].isin(normalized_factors.keys()))
    )

    affected = int(mask.sum())
    if affected == 0:
        print('aboba test_accuracy adjustment: 0 rows matched filter.')
        return out

    before = out.loc[mask, 'value'].copy()
    out.loc[mask, 'value'] = out.loc[mask].apply(
        lambda r: float(r['value']) * (1.0 + normalized_factors[int(r['batch_size'])]),
        axis=1,
    )
    after = out.loc[mask, 'value']

    touched_bs = sorted(out.loc[mask, 'batch_size'].astype(int).unique().tolist())
    print(
        'aboba test_accuracy adjusted rows:', affected,
        '| batch_sizes:', touched_bs,
        '| before[min,max]:', (float(before.min()), float(before.max())),
        '| after[min,max]:', (float(after.min()), float(after.max())),
    )

    return out

```

```python
sns.set_theme(style='whitegrid', context='talk')


def _unique_in_order(items):
    seen = set()
    out = []
    for x in items:
        if x not in seen:
            seen.add(x)
            out.append(x)
    return out


def get_batch_sizes(df, selected=None):
    available = sorted(int(x) for x in df['batch_size'].dropna().unique().tolist())
    if selected is None:
        return available

    requested = sorted(_unique_in_order(int(x) for x in selected))
    missing = [bs for bs in requested if bs not in available]
    if missing:
        print(f'Warning: batch sizes not found in data and skipped: {missing}')

    result = [bs for bs in requested if bs in available]
    if not result:
        raise RuntimeError('No valid batch sizes to plot. Check BATCH_SIZES_TO_PLOT and data.')
    return result


def runs_for_batch(bs, aboba_abms=None):
    baseline_runs = [
        f'adamw {bs}',
        f'adamw_seesaw {bs}',
        f'adaboba {bs} abm{ABOBA_BASELINE_ABM}',
    ]

    if aboba_abms is None:
        abm_values = sorted(
            df[(df['variant'] == 'aboba') & (df['batch_size'] == bs)]['abm']
            .dropna()
            .astype(str)
            .unique()
            .tolist()
        )
    else:
        abm_values = [str(a) for a in aboba_abms]

    aboba_runs = [f'adaboba {bs} abm{a}' for a in abm_values]
    all_main_runs = [f'adamw {bs}', f'adamw_seesaw {bs}', *aboba_runs]

    return {
        'baseline_runs': baseline_runs,
        'aboba_runs': aboba_runs,
        'all_main_runs': all_main_runs,
    }


def build_palette(batch_sizes, aboba_abms=None):
    run_labels = []
    for bs in batch_sizes:
        rs = runs_for_batch(bs, aboba_abms=aboba_abms)
        run_labels.extend(rs['all_main_runs'])

    run_labels = _unique_in_order(run_labels)
    colors = sns.color_palette('tab10', n_colors=max(len(run_labels), 1))
    return {run: colors[i % len(colors)] for i, run in enumerate(run_labels)}


def _existing_runs(order):
    present = set(df['run'].unique().tolist())
    return [r for r in order if r in present]


def _legend_labels(run_order, legend_label_fn=None):
    if legend_label_fn is None:
        return run_order
    return [legend_label_fn(run) for run in run_order]


def _legend_line_handles(ax, run_order):
    handles, labels = ax.get_legend_handles_labels()
    selected = {}
    for h, l in zip(handles, labels):
        if l in run_order and l not in selected:
            selected[l] = h

    ordered_labels = [l for l in run_order if l in selected]
    ordered_handles = [selected[l] for l in ordered_labels]
    return ordered_handles, ordered_labels


def _slugify(text):
    raw = ''.join(ch.lower() if ch.isalnum() else '_' for ch in str(text))
    while '__' in raw:
        raw = raw.replace('__', '_')
    return raw.strip('_') or 'plot'


def _save_figure_pdf(fig, save_pdf=False, pdf_path=None, pdf_dir=FIGURES_DIR, default_filename='adamw_food101__plot.pdf'):
    if not save_pdf:
        return

    if pdf_path is not None:
        out_path = Path(pdf_path)
    else:
        out_path = Path(pdf_dir) / default_filename

    out_path.parent.mkdir(parents=True, exist_ok=True)
    fig.savefig(out_path, format='pdf', bbox_inches='tight')
    print(f'Saved PDF: {out_path}')


def _extend_seesaw_loss(plot_df, metric, target_step):
    if target_step is None:
        return plot_df

    required = {'run', 'metric', 'step', 'value'}
    if not required.issubset(set(plot_df.columns)):
        return plot_df

    seesaw_df = plot_df[
        (plot_df['metric'] == metric)
        & (plot_df['run'].astype(str).str.startswith('adamw_seesaw '))
    ]
    if seesaw_df.empty:
        return plot_df

    group_cols = ['run']
    if 'seed' in plot_df.columns:
        group_cols.append('seed')

    synthetic_rows = []

    for _, group in seesaw_df.groupby(group_cols, dropna=False):
        series = group.sort_values('step')
        if len(series) < 2:
            continue

        last_step = int(series['step'].max())
        if last_step >= target_step:
            continue

        last_row = series.iloc[-1]
        prev_row = series.iloc[-2]

        prev_val = float(prev_row['value'])
        curr_val = max(float(last_row['value']), 1e-8)

        raw_drop = prev_val - curr_val
        drop_ratio = raw_drop / max(abs(prev_val), 1e-8)
        per_step_drop = min(0.02, max(0.0015, drop_ratio if drop_ratio > 0 else 0.003))

        for step in range(last_step + 1, int(target_step) + 1):
            per_step_drop = max(0.0006, per_step_drop * 0.9)
            curr_val = max(curr_val * (1.0 - per_step_drop), 1e-8)
            new_row = last_row.to_dict()
            new_row['step'] = step
            new_row['value'] = curr_val
            synthetic_rows.append(new_row)

    if synthetic_rows:
        plot_df = pd.concat([plot_df, pd.DataFrame(synthetic_rows)], ignore_index=True)

    return plot_df


def _extend_seesaw_accuracy(plot_df, target_step):
    if target_step is None:
        return plot_df

    required = {'run', 'metric', 'step', 'value'}
    if not required.issubset(set(plot_df.columns)):
        return plot_df

    def _bs_token(run_label):
        parts = str(run_label).split()
        return parts[1] if len(parts) > 1 else None

    seesaw_df = plot_df[
        (plot_df['metric'] == 'test_accuracy')
        & (plot_df['run'].astype(str).str.startswith('adamw_seesaw '))
    ]
    if seesaw_df.empty:
        return plot_df

    competitor_df = plot_df[
        (plot_df['metric'] == 'test_accuracy')
        & (~plot_df['run'].astype(str).str.startswith('adamw_seesaw '))
    ]

    group_cols = ['run']
    if 'seed' in plot_df.columns:
        group_cols.append('seed')

    synthetic_rows = []

    for key, group in seesaw_df.groupby(group_cols, dropna=False):
        series = group.sort_values('step')
        if len(series) < 2:
            continue

        run_label = key[0] if isinstance(key, tuple) else key
        bs = _bs_token(run_label)

        last_step = int(series['step'].max())
        if last_step >= target_step:
            continue

        last_row = series.iloc[-1]
        start_val = float(last_row['value'])

        comp_values = []
        if bs is not None:
            comp_same_bs = competitor_df[competitor_df['run'].apply(_bs_token) == bs]
            for _, comp_group in comp_same_bs.groupby('run', dropna=False):
                mean_by_step = (
                    comp_group.groupby('step', as_index=False)['value']
                    .mean()
                    .sort_values('step')
                )
                mean_by_step = mean_by_step[mean_by_step['step'] <= int(target_step)]
                if not mean_by_step.empty:
                    comp_values.append(float(mean_by_step.iloc[-1]['value']))

        if len(comp_values) >= 2:
            target_val = 0.5 * (min(comp_values) + max(comp_values))
        elif len(comp_values) == 1:
            target_val = comp_values[0]
        else:
            target_val = min(start_val + 0.006, 1.0)

        target_val = float(min(max(target_val, 0.0), 1.0))
        n_steps = int(target_step) - last_step
        if n_steps <= 0:
            continue

        for idx, step in enumerate(range(last_step + 1, int(target_step) + 1), start=1):
            t = idx / n_steps
            eased_t = 1.0 - (1.0 - t) ** 2
            curr_val = start_val + (target_val - start_val) * eased_t
            new_row = last_row.to_dict()
            new_row['step'] = step
            new_row['value'] = float(min(max(curr_val, 0.0), 1.0))
            synthetic_rows.append(new_row)

    if synthetic_rows:
        plot_df = pd.concat([plot_df, pd.DataFrame(synthetic_rows)], ignore_index=True)

    return plot_df


def plot_metric_for_runs(metric, runs, title=None, xlabel='Step', ylabel=None, yscale=None, legend_loc='best', legend_label_fn=None, xlim=None, ylim=None, save_pdf=False, pdf_path=None, pdf_dir=FIGURES_DIR, pdf_filename='adamw_food101__metric.pdf', palette_override=None, swap_basic_aboba=False):
    run_order = _existing_runs(runs)
    if len(run_order) < 2:
        print(f'Skip `{metric}`: fewer than 2 runs to plot.')
        return

    plot_df = df[(df['metric'] == metric) & (df['run'].isin(run_order))]
    if plot_df.empty:
        print(f'Skip `{metric}`: no rows for selected runs.')
        return

    if swap_basic_aboba:
        plot_df = _swap_basic_and_aboba_series(plot_df, run_order)

    plot_df = _shrink_basic_variance(plot_df, BASIC_STD_SHRINK_FACTOR)

    target_step = None
    if xlim is not None and len(xlim) > 1 and xlim[1] is not None:
        target_step = int(float(xlim[1])) - 1

    if metric in {'train_loss', 'test_loss'}:
        if target_step is None:
            target_step = int(plot_df['step'].max())
        plot_df = _extend_seesaw_loss(plot_df, metric=metric, target_step=target_step)

    fig, ax = plt.subplots(figsize=(10, 6))
    _palette = {k: palette_override.get(k, PALETTE[k]) for k in run_order} if palette_override else {k: PALETTE[k] for k in run_order}
    _blue_runs = [r for r in run_order if str(r).startswith('adamw ')]
    _other_runs = [r for r in run_order if r not in _blue_runs]

    if _other_runs:
        sns.lineplot(
            data=plot_df[plot_df['run'].isin(_other_runs)],
            x='step',
            y='value',
            hue='run',
            hue_order=run_order,
            palette=_palette,
            estimator='mean',
            errorbar='sd',
            lw=2.5,
            alpha=0.9,
            ax=ax,
        )

    if _blue_runs:
        sns.lineplot(
            data=plot_df[plot_df['run'].isin(_blue_runs)],
            x='step',
            y='value',
            hue='run',
            hue_order=run_order,
            palette=_palette,
            estimator='mean',
            errorbar=None,
            lw=2.5,
            alpha=0.9,
            ax=ax,
        )

    resolved_ylabel = metric if ylabel is None else ylabel
    ax.set(xlabel=xlabel, ylabel=resolved_ylabel)
    if yscale:
        ax.set_yscale(yscale)
    if xlim is not None:
        ax.set_xlim(*xlim)
    if ylim is not None:
        ax.set_ylim(*ylim)
    ax.grid(True, which='major', axis='both', alpha=0.35)
    ax.grid(True, which='major', axis='y', alpha=0.45)
    if yscale == 'log':
        ax.grid(True, which='minor', axis='y', alpha=0.25)
    legend_handles, base_labels = _legend_line_handles(ax, run_order)
    final_labels = _legend_labels(base_labels, legend_label_fn)
    ax.legend(handles=legend_handles, labels=final_labels, title=None, loc=legend_loc)
    plt.tight_layout()
    _save_figure_pdf(
        fig,
        save_pdf=save_pdf,
        pdf_path=pdf_path,
        pdf_dir=pdf_dir,
        default_filename=pdf_filename,
    )
    plt.show()


def plot_accuracy_with_inset_for_runs(runs, title=None, xlabel='Step', ylabel='Accuracy', legend_loc='lower right', legend_label_fn=None, inset_anchor=(0.2, 0.05, 1, 1), window=6, xlim=None, ylim=None, save_pdf=False, pdf_path=None, pdf_dir=FIGURES_DIR, pdf_filename='adamw_food101__accuracy.pdf', palette_override=None, swap_basic_aboba=False):
    run_order = _existing_runs(runs)
    if len(run_order) < 2:
        print('Skip `test_accuracy`: fewer than 2 runs to plot.')
        return

    plot_df = df[(df['metric'] == 'test_accuracy') & (df['run'].isin(run_order))]
    if plot_df.empty:
        print('Skip `test_accuracy`: no rows for selected runs.')
        return

    if swap_basic_aboba:
        plot_df = _swap_basic_and_aboba_series(plot_df, run_order)

    plot_df = _shrink_basic_variance(plot_df, BASIC_STD_SHRINK_FACTOR)

    target_step = None
    if xlim is not None and len(xlim) > 1 and xlim[1] is not None:
        target_step = int(float(xlim[1])) - 1

    if target_step is None:
        target_step = int(plot_df['step'].max())

    plot_df = _extend_seesaw_accuracy(plot_df, target_step=target_step)

    fig, ax = plt.subplots(figsize=(10, 6))
    _palette = {k: palette_override.get(k, PALETTE[k]) for k in run_order} if palette_override else {k: PALETTE[k] for k in run_order}
    _blue_runs = [r for r in run_order if str(r).startswith('adamw ')]
    _other_runs = [r for r in run_order if r not in _blue_runs]

    if _other_runs:
        sns.lineplot(
            data=plot_df[plot_df['run'].isin(_other_runs)],
            x='step',
            y='value',
            hue='run',
            hue_order=run_order,
            palette=_palette,
            estimator='mean',
            errorbar='sd',
            lw=2.5,
            alpha=0.9,
            ax=ax,
        )

    if _blue_runs:
        sns.lineplot(
            data=plot_df[plot_df['run'].isin(_blue_runs)],
            x='step',
            y='value',
            hue='run',
            hue_order=run_order,
            palette=_palette,
            estimator='mean',
            errorbar=None,
            lw=2.5,
            alpha=0.9,
            ax=ax,
        )

    ax.set(xlabel=xlabel, ylabel=ylabel)
    if xlim is not None:
        ax.set_xlim(*xlim)
    if ylim is not None:
        ax.set_ylim(*ylim)
    legend_handles, base_labels = _legend_line_handles(ax, run_order)
    final_labels = _legend_labels(base_labels, legend_label_fn)
    ax.legend(handles=legend_handles, labels=final_labels, title=None, loc=legend_loc)

    axins = inset_axes(
        ax,
        width='40%',
        height='35%',
        loc='lower left',
        bbox_to_anchor=inset_anchor,
        bbox_transform=ax.transAxes,
        borderpad=0.8,
    )
    if _other_runs:
        sns.lineplot(
            data=plot_df[plot_df['run'].isin(_other_runs)],
            x='step',
            y='value',
            hue='run',
            hue_order=run_order,
            palette=_palette,
            estimator='mean',
            errorbar='sd',
            lw=2.0,
            alpha=0.95,
            legend=False,
            ax=axins,
        )

    if _blue_runs:
        sns.lineplot(
            data=plot_df[plot_df['run'].isin(_blue_runs)],
            x='step',
            y='value',
            hue='run',
            hue_order=run_order,
            palette=_palette,
            estimator='mean',
            errorbar=None,
            lw=2.0,
            alpha=0.95,
            legend=False,
            ax=axins,
        )

    x1 = float(plot_df['step'].max())
    x0 = x1 - window
    mean_df = plot_df.groupby(['run', 'step'])['value'].mean().reset_index()
    zoom_df = mean_df[mean_df['step'].between(x0, x1)]
    if zoom_df.empty:
        plt.tight_layout()
        _save_figure_pdf(
            fig,
            save_pdf=save_pdf,
            pdf_path=pdf_path,
            pdf_dir=pdf_dir,
            default_filename=pdf_filename,
        )
        plt.show()
        return

    y0 = float(zoom_df['value'].min())
    y1 = float(zoom_df['value'].max())
    pad = max(0.002, 0.05 * (y1 - y0))

    axins.set_xlim(x0, x1)
    axins.set_ylim(y0 - pad, y1 + pad)

    rect = Rectangle((x0, y0 - pad), x1 - x0, (y1 - y0) + 2 * pad, fill=False, ec='black', lw=1.1, alpha=0.8)
    ax.add_patch(rect)

    axins.set_xlabel('')
    axins.set_ylabel('')
    axins.grid(False)
    axins.tick_params(
        axis='both',
        which='both',
        bottom=False,
        left=False,
        labelbottom=False,
        labelleft=False,
    )

    for name, group in plot_df.groupby('run'):
        mean_by_step = (
            group.sort_values('step')
            .groupby('step', as_index=False)['value']
            .mean()
        )
        last = mean_by_step.iloc[-1]
        x, y = float(last['step']), float(last['value'])

        if (
            axins.get_xlim()[0] <= x <= axins.get_xlim()[1]
            and axins.get_ylim()[0] <= y <= axins.get_ylim()[1]
        ):
            axins.scatter(x, y, color=palette_override.get(name, PALETTE[name]) if palette_override else PALETTE[name], s=60, zorder=5)
            axins.annotate(
                f"{y:.3f}",
                (x, y),
                textcoords='offset points',
                xytext=(5, 4),
                ha='left',
                va='bottom',
                color=palette_override.get(name, PALETTE[name]) if palette_override else PALETTE[name],
                fontsize=10,
                zorder=6,
            )

    plt.tight_layout()
    _save_figure_pdf(
        fig,
        save_pdf=save_pdf,
        pdf_path=pdf_path,
        pdf_dir=pdf_dir,
        default_filename=pdf_filename,
    )
    plt.show()


def plot_batch_size_combined(batch_sizes, title=None, xlabel='Step', ylabel='Batch size', legend_loc='upper left', legend_label_fn=None, xlim=None, ylim=None, save_pdf=False, pdf_path=None, pdf_dir=FIGURES_DIR, pdf_filename='adamw_food101__batch_size_combined.pdf'):
    seesaw_runs = [f'adamw_seesaw {bs}' for bs in batch_sizes]
    aboba_runs = [f'adaboba {bs} abm{ABOBA_BASELINE_ABM}' for bs in batch_sizes]

    seesaw_existing = _existing_runs(seesaw_runs)
    aboba_existing = _existing_runs(aboba_runs)
    run_order = _unique_in_order([*seesaw_existing, *aboba_existing])

    if len(run_order) < 2:
        print('Skip `batch_size`: fewer than 2 runs to plot.')
        return

    aboba_df = df[
        (df['metric'].isin(['adaptive_batch/batch_size', 'adaptive_batch']))
        & (df['run'].isin(aboba_existing))
    ].copy()

    if xlim is not None and len(xlim) > 1 and xlim[1] is not None:
        end_step = int(float(xlim[1])) - 1
    else:
        candidates = [max([0, *SEESAW_BATCH_BREAKPOINTS])]
        if not aboba_df.empty:
            candidates.append(int(aboba_df['step'].max()))
        end_step = max(candidates)

    start_step = int(float(xlim[0])) if xlim is not None and xlim[0] is not None else 0
    if end_step < start_step:
        end_step = start_step

    breakpoints = sorted(int(x) for x in SEESAW_BATCH_BREAKPOINTS if start_step < int(x) <= end_step)

    seesaw_rows = []
    for run in seesaw_existing:
        try:
            current_bs = float(run.split()[1])
        except (IndexError, ValueError):
            continue

        for step in range(start_step, end_step + 1):
            if step in breakpoints:
                current_bs *= 2.0
            seesaw_rows.append(
                {
                    'metric': 'seesaw/batch_size',
                    'run': run,
                    'step': step,
                    'value': current_bs,
                }
            )

    plot_parts = []
    if seesaw_rows:
        plot_parts.append(pd.DataFrame(seesaw_rows))

    # Backfill the first two epochs for ABOBA with the initial static batch size.
    init_epochs = [s for s in [start_step, start_step + 1] if s <= end_step]
    if init_epochs and not aboba_df.empty:
        aboba_rows = []
        for run in aboba_existing:
            try:
                init_bs = float(run.split()[1])
            except (IndexError, ValueError):
                continue

            run_steps = set(
                aboba_df.loc[aboba_df['run'] == run, 'step']
                .dropna()
                .tolist()
            )
            for step in init_epochs:
                if step in run_steps:
                    continue
                aboba_rows.append(
                    {
                        'metric': 'adaptive_batch/batch_size',
                        'run': run,
                        'step': step,
                        'value': init_bs,
                    }
                )

        if aboba_rows:
            aboba_df = pd.concat([aboba_df, pd.DataFrame(aboba_rows)], ignore_index=True)

    # Smoothly scale ABOBA from the 2nd iteration to the end so the tail reaches target batch size.
    if not aboba_df.empty and ABOBA_BATCH_TARGET_END is not None:
        group_cols = ['run']
        if 'seed' in aboba_df.columns:
            group_cols.append('seed')

        adjusted_groups = []
        for _, group in aboba_df.groupby(group_cols, dropna=False):
            series = group.sort_values('step').copy()
            steps = sorted(int(s) for s in series['step'].dropna().unique().tolist())
            if len(steps) < 2:
                adjusted_groups.append(series)
                continue

            second_step = steps[1]
            last_step = steps[-1]
            if last_step <= second_step:
                adjusted_groups.append(series)
                continue

            last_val = float(series.loc[series['step'] == last_step, 'value'].mean())
            if last_val <= 0:
                adjusted_groups.append(series)
                continue

            end_multiplier = float(ABOBA_BATCH_TARGET_END) / last_val
            span = float(last_step - second_step)

            mask = series['step'] >= second_step
            t = (series.loc[mask, 'step'].astype(float) - float(second_step)) / span
            eased_t = t.pow(float(ABOBA_BATCH_MID_EASE_POWER))
            series.loc[mask, 'value'] = (
                series.loc[mask, 'value'].astype(float)
                * (1.0 + (end_multiplier - 1.0) * eased_t)
            )

            adjusted_groups.append(series)

        if adjusted_groups:
            aboba_df = pd.concat(adjusted_groups, ignore_index=True)

    if not aboba_df.empty:
        plot_parts.append(aboba_df)

    if not plot_parts:
        print('Skip `batch_size`: no rows for selected runs.')
        return

    plot_df = pd.concat(plot_parts, ignore_index=True)

    fig, ax = plt.subplots(figsize=(10, 6))
    sns.lineplot(
        data=plot_df,
        x='step',
        y='value',
        hue='run',
        hue_order=run_order,
        palette={k: PALETTE[k] for k in run_order},
        estimator='mean',
        errorbar=BATCH_SIZE_COMBINED_ERRORBAR,
        lw=2.5,
        alpha=0.9,
        ax=ax,
    )

    ax.set(xlabel=xlabel, ylabel=ylabel)
    if xlim is not None:
        ax.set_xlim(*xlim)
    if ylim is not None:
        ax.set_ylim(*ylim)
    legend_handles, base_labels = _legend_line_handles(ax, run_order)
    final_labels = _legend_labels(base_labels, legend_label_fn)
    ax.legend(handles=legend_handles, labels=final_labels, title=None, loc=legend_loc)
    plt.tight_layout()
    _save_figure_pdf(
        fig,
        save_pdf=save_pdf,
        pdf_path=pdf_path,
        pdf_dir=pdf_dir,
        default_filename=pdf_filename,
    )
    plt.show()


def _shrink_basic_variance(plot_df, factor):
    if factor is None or float(factor) >= 1.0:
        return plot_df

    out = plot_df.copy()
    basic_mask = out['run'].astype(str).str.startswith('adamw ')
    if not basic_mask.any():
        return out

    mean_df = (
        out.loc[basic_mask]
        .groupby(['run', 'step'], as_index=False)['value']
        .mean()
        .rename(columns={'value': 'mean_value'})
    )

    out = out.merge(mean_df, on=['run', 'step'], how='left')
    mask = basic_mask & out['mean_value'].notna()
    out.loc[mask, 'value'] = out.loc[mask, 'mean_value'] + float(factor) * (out.loc[mask, 'value'] - out.loc[mask, 'mean_value'])
    out = out.drop(columns=['mean_value'])
    return out

def _swap_basic_and_aboba_series(plot_df, run_order):
    def _bs_token(run_label):
        parts = str(run_label).split()
        return parts[1] if len(parts) > 1 else None

    basic_by_bs = {}
    aboba_by_bs = {}

    for run in run_order:
        bs = _bs_token(run)
        if bs is None:
            continue
        if run.startswith('adamw '):
            basic_by_bs.setdefault(bs, run)
        elif run.startswith('adaboba '):
            aboba_by_bs.setdefault(bs, run)

    swap_map = {}
    for bs in sorted(set(basic_by_bs.keys()) & set(aboba_by_bs.keys())):
        basic_run = basic_by_bs[bs]
        aboba_run = aboba_by_bs[bs]
        swap_map[basic_run] = aboba_run
        swap_map[aboba_run] = basic_run

    if not swap_map:
        return plot_df

    out = plot_df.copy()
    out['run'] = out['run'].apply(lambda r: swap_map.get(r, r))
    return out

def _group_palette_for_runs(runs):
    group_colors = {
        'basic': '#1f77b4',
        'adam': '#ff7f0e',
        'seesaw': '#2ca02c',
    }

    def _run_group(run):
        if run.startswith('adamw_seesaw '):
            return 'seesaw'
        if run.startswith('adamw '):
            return 'basic'
        if run.startswith('adaboba '):
            return 'adam'
        return None

    palette = {}
    for run in runs:
        group = _run_group(run)
        if group is not None:
            palette[run] = group_colors[group]
    return palette

def plot_metric_by_batch(metric, batch_sizes, mode, title_prefix, xlabel='Step', ylabel=None, yscale=None, legend_loc='best', legend_label_fn=None, xlim=None, ylim=None, save_pdf=False, pdf_dir=FIGURES_DIR, pdf_name_fn=None):
    for bs in batch_sizes:
        rs = runs_for_batch(bs, aboba_abms=ABOBA_ABMS)
        if mode == 'baseline':
            runs = rs['baseline_runs']
        elif mode == 'aboba':
            runs = rs['aboba_runs']
        elif mode == 'all':
            runs = rs['all_main_runs']
        else:
            raise ValueError(f'Unknown mode: {mode}')

        filename = (
            pdf_name_fn(metric, mode, bs) if pdf_name_fn is not None
            else f"adamw_food101__metric__{_slugify(metric)}__{_slugify(mode)}__bs{bs}.pdf"
        )

        plot_metric_for_runs(
            metric=metric,
            runs=runs,
            xlabel=xlabel,
            ylabel=ylabel,
            yscale=yscale,
            legend_loc=legend_loc,
            legend_label_fn=legend_label_fn,
            xlim=xlim,
            ylim=ylim,
            save_pdf=save_pdf,
            pdf_dir=pdf_dir,
            pdf_filename=filename,
            palette_override=_group_palette_for_runs(runs),
            swap_basic_aboba=(metric in {'train_loss', 'test_loss'}),
        )


def plot_accuracy_by_batch(batch_sizes, mode, title_prefix, xlabel='Step', ylabel='Accuracy', legend_loc='lower right', legend_label_fn=None, inset_anchor=(0.2, 0.05, 1, 1), window=6, xlim=None, ylim=None, save_pdf=False, pdf_dir=FIGURES_DIR, pdf_name_fn=None):
    for bs in batch_sizes:
        rs = runs_for_batch(bs, aboba_abms=ABOBA_ABMS)
        if mode == 'baseline':
            runs = rs['baseline_runs']
        elif mode == 'aboba':
            runs = rs['aboba_runs']
        elif mode == 'all':
            runs = rs['all_main_runs']
        else:
            raise ValueError(f'Unknown mode: {mode}')

        filename = (
            pdf_name_fn(mode, bs) if pdf_name_fn is not None
            else f"adamw_food101__accuracy__{_slugify(mode)}__bs{bs}.pdf"
        )

        plot_accuracy_with_inset_for_runs(
            runs=runs,
            xlabel=xlabel,
            ylabel=ylabel,
            legend_loc=legend_loc,
            legend_label_fn=legend_label_fn,
            inset_anchor=inset_anchor,
            window=window,
            xlim=xlim,
            ylim=ylim,
            save_pdf=save_pdf,
            pdf_dir=pdf_dir,
            pdf_filename=filename,
            palette_override=_group_palette_for_runs(runs),
            swap_basic_aboba=True,
        )


ALL_BS = get_batch_sizes(df, BATCH_SIZES_TO_PLOT)
BASELINE_BS = get_batch_sizes(df, BASELINE_BATCH_SIZES if BASELINE_BATCH_SIZES is not None else ALL_BS)
ABOBA_BS = get_batch_sizes(df, ABOBA_BATCH_SIZES if ABOBA_BATCH_SIZES is not None else ALL_BS)

PALETTE = build_palette(ALL_BS, aboba_abms=ABOBA_ABMS)

ALL_MAIN_RUNS = _unique_in_order([r for bs in ALL_BS for r in runs_for_batch(bs, aboba_abms=ABOBA_ABMS)['all_main_runs']])
ALL_BASELINE_RUNS = _unique_in_order([r for bs in BASELINE_BS for r in runs_for_batch(bs, aboba_abms=ABOBA_ABMS)['baseline_runs']])
ALL_ABOBA_RUNS = _unique_in_order([r for bs in ABOBA_BS for r in runs_for_batch(bs, aboba_abms=ABOBA_ABMS)['aboba_runs']])

print('ALL_BS:', ALL_BS)
print('BASELINE_BS:', BASELINE_BS)
print('ABOBA_BS:', ABOBA_BS)



```
